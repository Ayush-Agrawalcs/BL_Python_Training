```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

# Write a Python programming to display a bar chart of the popularity of programming Languages.

```python
programming=['Java', 'Python', 'PHP', 'JavaScript',' C#', 'C++']
popularity=[22.2, 17.6, 8.8, 8, 7.7, 6.7]
plt.bar(programming,popularity)
plt.show()
```

# Write a Python programming to display a horizontal bar chart of the popularity ofcprogramming Languages.

```python
programming=['Java', 'Python', 'PHP', 'JavaScript',' C#', 'C++']
popularity=[22.2, 17.6, 8.8, 8, 7.7, 6.7]
plt.barh(programming,popularity)
plt.show()
```

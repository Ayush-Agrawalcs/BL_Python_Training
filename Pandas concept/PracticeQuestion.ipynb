```python
import pandas as pd
import numpy as np

np.random.seed(42)
n = 500

df = pd.DataFrame({
    'order_id': np.arange(1, n+1),
    'order_date': pd.date_range(start='2023-01-01', periods=n, freq='D'),
    'customer_id': np.random.randint(1000, 1100, n),
    'product_category': np.random.choice(
        ['Electronics', 'Clothing', 'Home', 'Sports', 'Books'], n
    ),
    'price': np.round(np.random.uniform(100, 1000, n), 2),
    'quantity': np.random.randint(1, 5, n),
    'discount_percent': np.random.choice([0, 5, 10, 15, 20], n)
})

# Add outliers
outlier_idx = np.random.choice(df.index, 10)
df.loc[outlier_idx, 'price'] *= 5
```

```python
df.head()
```

```python
df['final_amount']=(df['price']*df['quantity'])*(1-df['discount_percent']/100)
```

```python
df.head()
```

```python
df['month']=df['order_date'].dt.to_period('M')
df
```

```python
df['monthly_revenue'] = df.groupby(['month', 'product_category'])['final_amount'].transform('sum')
```

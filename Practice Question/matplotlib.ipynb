```python
data={
    "salary":[25000,3000,37000,28000,39000,48000,55000,52000,35000]
}
```

```python
df=pd.DataFrame(data)
```

```python
df.head()
```

```python
df.shape
```

```python
#Line plot
plt.plot(df["salary"],color="red",marker="*",linestyle=":",linewidth="2")
plt.grid()
plt.show()
```

```python
# Histogram
plt.hist(df["salary"],bins=5,color="green")
plt.show()
```

```python
#box plot
plt.boxplot(df["salary"])
plt.show()
```

```python
df["dept"]=["HR","IT","Finance","HR","Finance","HR","IT","Finance","HR","IT"]
```

```python
df.head()
```

```python
df.info()
```

```python
#pie chart
count=df["dept"].value_counts()
count
```

```python
plt.pie(count,labels=count.index,autopct="%1.1f",explode=[0,0.1,0.2])
plt.axis("equal")
plt.show()
```

```python
#count plot
plt.bar(count.index,count,color=["green","black","red"])
plt.show()
```

# Write a Python program to draw a line using given axis values taken from a text file, with suitable label in the x axis, y axis and a title.

```python
x = []
y = []

with open("test.txt", "r") as f:
    for line in f:
        v = line.split()
        x.append(float(v[0]))  
        y.append(float(v[1])) 
        

plt.plot(x, y)
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot from File')
plt.show()
```

# Write a Python program to draw line charts of the financial data of Alphabet Inc. between October 3, 2016 to October 7, 2016.

```python
x=[]
y1=[]
y2=[]
y3=[]
y4=[]
count=0
with open("fdata.csv", "r") as f:
    for line in f:
        v = line.split(",")
        if(count==0):
            count+=1
            continue
        x.append(v[0])  
        y1.append(float(v[1]))
        y2.append(float(v[2])) 
        y3.append(float(v[3])) 
        y4.append(float(v[4])) 
plt.plot(x,y1)
plt.plot(x,y2)
plt.plot(x,y3)
plt.plot(x,y4)
plt.xlabel("Date")
plt.ylabel("financial data of Alphabet")
plt.show()
```

# Write a Python program to plot several lines with different format styles in one command using arrays.

```python
import matplotlib.pyplot as plt
import numpy as np

# Create arrays
x = np.array([1, 2, 3, 4, 5])
y1 = x
y2 = x ** 2
y3 = x ** 3


plt.plot(x, y1, 'r--',   
         x, y2, 'bs-',   
         x, y3, 'g^:')   

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Multiple Lines with Different Format Styles")

plt.legend(["y = x", "y = x^2", "y = x^3"])

plt.show()
```

# Write a Python program to create multiple types of charts

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
categories=['A','B','C','D','E']
values=[40,50,60,50,80]

#  Line chart
plt.figure()
plt.plot(x, y)
plt.title("Line Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# Br chart
plt.figure()
plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()

# 3️⃣ Scatter Plot
plt.figure()
plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()


# 4️⃣ Pie Chart
plt.figure()
plt.pie(values, labels=categories, autopct='%1.1f%%')
plt.title("Pie Chart")
plt.show()
```

# Write a Python program to display the grid and draw line charts of the closing value of Alphabet Inc. between October 3, 2016 to October 7, 2016. Customized the grid lines with rendering with a larger grid (major grid) and a smaller grid (minor grid).Turn on the grid but turn off ticks. 

```python
data=pd.read_csv('closing.csv')
x=data['Date']
y=data['closing']

plt.plot(x,y,linestyle='-',linewidth="5")
plt.grid()
plt.xlabel("Date")
plt.ylabel("Closing")
plt.title("Alphabet Inc. Closing Prices")
plt.grid(which='major', linestyle='-', linewidth=1)

plt.minorticks_on()
plt.grid(which='minor', linestyle=':', linewidth=0.5)

plt.show()
```

# Write a Python program to create multiple plots.

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [1, 4, 9, 16, 25]

plt.subplot(2, 1, 1)   
plt.plot(x, y1)
plt.title("First Plot")
plt.xlabel("X-axis")
plt.ylabel("Y1 values")

plt.subplot(2, 1, 2)   
plt.plot(x, y2)
plt.title("Second Plot")
plt.xlabel("X-axis")
plt.ylabel("Y2 values")

plt.tight_layout()  
plt.show()
```

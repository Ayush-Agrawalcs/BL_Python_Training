# practice Question on Data Visualization

```python
programming=['java','python','php','javascript','c#','c++']
popularity=[22.2,15.6,8.8,7.7,6.7,8.5]
plt.pie(popularity,labels=programming,autopct='%1.1f%%')
plt.show()
```

# Write a Python programming to create a pie chart with a title of the popularity of programming Languages.

```python
programming=['java','python','php','javascript','c#','c++']
popularity=[22.2,15.6,8.8,7.7,6.7,8.5]
plt.pie(popularity,labels=programming,autopct='%1.1f%%')
plt.title("Popularity of Programming Languages")
plt.show()
```

# Write a Python programming to create a pie chart of gold medal achievements of five mostsuccessful countries in 2016 Summer Olympics. Read the data from a csv file. Sample data:

```python
data=pd.read_csv('Olympics.csv')
country=data['country']
medal=data['gold_medal']
plt.pie(medal,labels=country,autopct="%1.1f%%")
plt.show()
```
